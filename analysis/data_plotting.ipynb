# Data plotting


Si effettua uno scatter plot con le colonne P e A, queste due variabili (P e A) sono spesso fondamentali per caratterizzare diversi tipi di stelle variabili, ed è per questo che vengono messe in relazione in un grafico.

- **Periodo (P)**:

  Il periodo (P) di una stella variabile è il tempo impiegato dalla stella per completare un ciclo di variazione della sua luminosità. 
  È una caratteristica fondamentale per distinguere diversi tipi di stelle variabili. Alcune classi di stelle variabili hanno periodi molto lunghi, mentre altre mostrano variazioni rapide.
- **Ampiezza (A)**:

  L'ampiezza (A) si riferisce alla differenza tra la luminosità massima e minima della stella. Anche questa è una caratteristica importante per determinare il tipo di stella variabile, poiché diverse classi di stelle variabili possono avere ampiezze diverse (cioè differenze più o meno grandi nella variazione di luminosità).

## Perché P e A insieme?

**Relazione fisica**: C'è una relazione tra il periodo e l'ampiezza che può aiutare a distinguere le diverse classi di stelle variabili. Ad esempio, stelle variabili di classi diverse tendono a raggrupparsi in zone diverse nel grafico periodo-ampiezza.

**Classificazione delle stelle variabili**: In astronomia, grafici di questo tipo sono spesso utilizzati per osservare pattern e separare visivamente le stelle in base al loro tipo. Osservando come si dispongono i punti delle diverse classi di stelle nel grafico, si possono individuare differenze nei loro periodi e nelle ampiezze che aiutano nella loro classificazione.
*In questo contesto, il grafico aiuta a visualizzare la distribuzione delle stelle variabili nello spazio delle due caratteristiche (P e A) e a vedere se esistono raggruppamenti distinti per i vari tipi di stelle variabili (LCtype)*.

## Perchè si realizza uno scatter plot?
**Relazione tra variabili**: Aiuta a identificare se c'è una relazione o correlazione tra le due variabili. Ad esempio, se i punti tendono a disporsi secondo una linea retta, potrebbe esserci una correlazione lineare tra le due variabili.

**Distribuzione**: Può mostrare la distribuzione dei dati e identificare eventuali gruppi o pattern, come cluster di punti che potrebbero indicare categorie naturali o tipi di oggetti (come nel caso delle stelle variabili).

**Outliers**: Permette di individuare eventuali punti anomali (outliers) che si discostano molto dalla distribuzione principale.

```python
import matplotlib.pyplot as plt
import pandas as pd
import os

# File path
parent_directory = os.path.dirname(os.getcwd())

# file_path = r"C:/Users/d.dipadova/ClassificationOfVariableStars/VariableStarClassification/data/output/output.csv"
file_path = parent_directory + r"\data\output\output.csv"

# Carica i dati
data = pd.read_csv(file_path, sep=",")

# Crea una figura
plt.figure(figsize=(7, 7))

# Itera su ciascun tipo di stella (LCtype)
for starType in data['LCtype'].unique():

    # Filtra il dataframe per la classe corrente
    subset = data[data['LCtype'] == starType]
    
    # Crea uno scatter plot per la classe corrente
    plt.scatter(subset['P'], subset['A'], label=starType, alpha=0.5) # alpha: introduce trasparenza 

# Aggiungi etichette e legenda
plt.xlabel('P')
plt.ylabel('A')
plt.legend(title='LCtype')
plt.title('Scatter plot delle diverse classi (LCtype)')
plt.show()
```

Estendiamo il data plotting ad ogni coppia di colonne del dataset, non limitandoci all'ampiezza (A) e al periodo (P)

```python
from itertools import combinations

# 1. Ottieni le colonne del dataframe
colonne = data.columns

# 2. Crea tutte le combinazioni di coppie di colonne

coppie = list(combinations(colonne, 2))

coppie_list = list(coppie)



for i in range(1,6):
    for coppia in coppie_list:
        if( coppia[0] == "LINEARobjectID" or coppia[0] == "LCtype"  ):
            coppie_list.remove(coppia)
            

```

```python
for coppia in coppie_list:
    print(coppia)
    # Crea una figura
    plt.figure(figsize=(5, 5))

    # Itera su ciascun tipo di stella (LCtype)
    for starType in data['LCtype'].unique():
        
        # Filtra il dataframe per la classe corrente
        subset = data[data['LCtype'] == starType]
        
        # Crea uno scatter plot per la classe corrente
        plt.scatter(subset[coppia[0]], subset[coppia[1]], label=starType, alpha=0.7)

    # Aggiungi etichette e legenda
    plt.xlabel(coppia[0])
    plt.ylabel(coppia[1])
    plt.legend(title='LCtype')
    plt.title('Scatter plot delle diverse classi (LCtype)')
    plt.show()
```

# Analisi delle correlazioni

```python
# Calcolo della matrice di correlazione
correlation_matrix = data.corr()  # Usa .corr() per le correlazioni di Pearson

# Creazione della heatmap
plt.figure(figsize=(10, 8))  # Regola la dimensione della figura
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Titolo e visualizzazione
plt.title('Heatmap delle correlazioni')
plt.show()
```

Correlazioni positive e negative
Correlazione positiva:

Quando una variabile aumenta, anche l'altra aumenta.
Esempio: Altezza e peso tendono ad avere una correlazione positiva (più sei alto, più pesi, in media).
Correlazione negativa:

Quando una variabile aumenta, l'altra diminuisce.
Esempio: Il prezzo di un prodotto e la quantità venduta spesso hanno una correlazione negativa (all’aumentare del prezzo, le vendite diminuiscono).
